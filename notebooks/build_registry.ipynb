```python
import json
from pathlib import Path
from collections import defaultdict
import pandas as pd
```

```python
Tool_dir=Path("/home/abhishek/Desktop/projects/RAG_MOLE/data/raw/ToolBench/data/data/toolenv/tools")
```

```python
json_files = [p for p in Tool_dir.rglob("*.json") if "toolllama_G123" not in p.name]
print(f"Found {len(json_files)} JSON files.")
json_files[:20]
```

```python
for p in list(Tool_dir.iterdir())[:]:
    print(p.name)
```

```python
registry = []

for file_path in json_files:
    try:
        with open(file_path, "r", encoding="utf-8") as f:
            tool = json.load(f)

    except Exception as e:
        print(f"Skipping {file_path}: {e}")
        continue

    category = file_path.parent.name
    score = tool.get("score") or {}
    tool_name = tool.get("tool_name", "")
    tool_description = tool.get("tool_description", "")

    for api in tool.get("api_list", []):
        registry.append({
            # Identity
            "product_id": tool.get("product_id", ""),
            "category": category,
            "tool_name": tool_name,
            "tool_title": tool.get("title", ""),
            "api_name": api.get("name", ""),

            # Semantic Information
            "tool_description": tool_description,
            "api_description": api.get("description", ""),

            # Endpoint Information
            "url": api.get("url", ""),
            "method": api.get("method", ""),
            "required_parameters": api.get(
                "required_parameters", []
            ),
            "optional_parameters": api.get(
                "optional_parameters", []
            ),

            # Operational Metadata
            "pricing": tool.get("pricing", ""),
            "avg_latency": score.get(
                "avgLatency", None
            ),
            "avg_success_rate": score.get(
                "avgSuccessRate", None
            ),
            "avg_service_level": score.get(
                "avgServiceLevel", None
            ),
            "popularity_score": score.get(
                "popularityScore", None
            ),

            # Source Information
            "home_url": tool.get("home_url", ""),
            "host": tool.get("host", "")
        })

print(f"Registry Entries: {len(registry)}")
```

```python
df=pd.DataFrame(registry)
df.iloc[0:10,:]
```

This is simple Normalization of the data types of the parameters thats it


```python
TYPE_MAPPING = {
    "STRING": "STRING",
    "string": "STRING",

    "NUMBER": "NUMBER",
    "FLOAT": "NUMBER",

    "BOOLEAN": "BOOLEAN",

    "DATE": "DATE",
    "DATEPICKER": "DATE",
    "DATE (YYYY-MM-DD)": "DATE",

    "DATETIME": "DATETIME",

    "TIME (24-hour HH:MM)": "TIME",

    "ARRAY": "ARRAY",
    "LIST": "ARRAY",

    "OBJECT": "OBJECT",
    "JSON": "JSON",
    "MAP": "MAP",

    "FILE": "FILE",
    "BINARY": "FILE",

    "ENUM": "ENUM",
    "SELECT": "ENUM",

    "CREDENTIALS": "CREDENTIALS",

    "GEOPOINT (latitude, longitude)": "GEOPOINT",

    None: "UNKNOWN",
    "": "UNKNOWN"
}

def normalize_parameter_types(params, mapping):
    for param in params:
        original = param.get("type")
        param["type"] = mapping.get(original, original)

for row in registry:
    normalize_parameter_types(
        row["required_parameters"],
        TYPE_MAPPING
    )

    normalize_parameter_types(
        row["optional_parameters"],
        TYPE_MAPPING
    )
```

```python
df["num_required_params"] = df["required_parameters"].apply(len)

df["num_optional_params"] = df["optional_parameters"].apply(len)

df["requires_credentials"] = df["required_parameters"].apply(
    lambda params: any(
        p.get("type") == "CREDENTIALS"
        for p in params
    )
)
df.shape
```
