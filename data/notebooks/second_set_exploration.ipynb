```python
def summarize_dataset(name, df):
    print(f"\n{'=' * 80}")
    print(f"Dataset: {name}")
    print(f"{'=' * 80}")
    print("Shape:", df.shape)
    print("Columns:", list(df.columns))
    print("\nData types:")
    print(df.dtypes)
    print("\nMissing values:")
    print(df.isnull().sum())
    print("\nPreview:")
    display(df.head(3))
    print("\nDescriptive summary:")
    display(df.describe(include="all"))


for name, df in [
    ("tool_creation", df_tool_creation),
    ("tool_selection", df_tool_selection),
    ("tool_usage_awareness", df_tool_usage_awareness),
    ("tool_usage", df_tool_usage),
]:
    summarize_dataset(name, df)
```

```python
def inspect_column_types(name, df, max_rows=3):
    print(f"\n{'=' * 80}")
    print(f"Column inspection for {name}")
    print(f"{'=' * 80}")
    for col in df.columns:
        print(f"\nColumn: {col}")
        sample_vals = df[col].dropna().iloc[:max_rows].tolist()
        print("Sample values:", sample_vals)
        print("Type of first non-null value:", type(sample_vals[0]).__name__ if sample_vals else None)
        print("Null count:", df[col].isnull().sum())


for name, df in [
    ("tool_creation", df_tool_creation),
    ("tool_selection", df_tool_selection),
    ("tool_usage_awareness", df_tool_usage_awareness),
    ("tool_usage", df_tool_usage),
]:
    inspect_column_types(name, df)
```

```python
def find_surrogates(obj, path=""):
    if isinstance(obj, str):
        bad = [
            (i, hex(ord(c)), repr(c))
            for i, c in enumerate(obj)
            if 0xD800 <= ord(c) <= 0xDFFF
        ]

        if bad:
            return [(path, bad, obj)]

    elif isinstance(obj, dict):
        results = []
        for key, value in obj.items():
            results.extend(
                find_surrogates(value, f"{path}[{key!r}]")
            )
        return results

    elif isinstance(obj, (list, tuple)):
        results = []
        for i, value in enumerate(obj):
            results.extend(
                find_surrogates(value, f"{path}[{i}]")
            )
        return results

    return []
```

```python
for col in df.columns:
    for idx, value in df[col].items():
        results = find_surrogates(value, col)

        if results:
            print(f"\nRow: {idx}")
            for result in results:
                print(result)
```

```python
def fix_surrogates(obj):
    if isinstance(obj, str):
        return obj.encode("utf-16", "surrogatepass").decode("utf-16")

    elif isinstance(obj, dict):
        return {
            key: fix_surrogates(value)
            for key, value in obj.items()
        }

    elif isinstance(obj, list):
        return [
            fix_surrogates(value)
            for value in obj
        ]

    elif isinstance(obj, tuple):
        return tuple(
            fix_surrogates(value)
            for value in obj
        )

    return obj
```

```python
df = df.map(fix_surrogates)
```
